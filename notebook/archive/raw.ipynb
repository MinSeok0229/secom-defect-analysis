```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from pyprojroot import here
```

```python
DATA_DIR = here("data/raw")
print(DATA_DIR)   # ...\SECOM\data\raw 나오는지 확인

```

```python
# 피처 (공백 구분, 결측치는 NaN으로 들어있음)
X = pd.read_csv(DATA_DIR / "secom.data", sep=r"\s+", header=None)
X.columns = [f"f{i}" for i in range(X.shape[1])]
```

```python
# 라벨 + 타임스탬프 (예: -1 "19/07/2008 11:55:00")
y = pd.read_csv(DATA_DIR / "secom_labels.data", sep=" ", header=None,
                names=["label", "timestamp"], quotechar='"')
y["timestamp"] = pd.to_datetime(y["timestamp"], format="%d/%m/%Y %H:%M:%S")
```

```python
# 합치기
df = pd.concat([y, X], axis=1)
print(df.shape)                      # (1567, 592) 나오면 정상
print(df["label"].value_counts())
```

```python
na_ratio = X.isna().mean().sort_values(ascending=False)
print(na_ratio.describe())
print((na_ratio > 0.5).sum(), "개 컬럼이 결측 50% 초과")
```

```python
print('1개 값만 가지는 컬럼:')
print(df.columns[df.nunique() == 1])  # 1개 값만 가지는 컬럼 확인
print()
print('1개만 가지는 컬럼 갯수:', df.columns[df.nunique() == 1].shape[0])  
    
print()

print('1개이하 가지는 컬럼 갯수:', df.columns[df.nunique() <= 1].shape[0]) 


# 정보 특징 없음... just drop it
df = df.drop(columns=df.columns[df.nunique() <= 1])
print('df shape after drop 1개 이하 컬럼:', df.shape)   # col 590 -> 476
```

```python
print(f'양품 비율: {(df[df["label"] == -1].shape[0] / df.shape[0]):.4f}')  # 양품 비율 확인 # -1 pass/ 1 fail
```
